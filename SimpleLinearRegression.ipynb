```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("placement.csv")

print(df.head())
print(df.shape)
```

```python
X = df["cgpa"].values
y = df["placed"].values

print(X[:5])
print(y[:5])
```

```python
plt.scatter(X,y)
plt.xlabel("CGPA")
plt.ylabel("Placed")
plt.title("CGPA vs Placement")
plt.show()
```

```python
m = 0
b = 0

learning_rate = 0.01
epochs = 1000

n = len(X)
```

```python
cost_history = []

for i in range(epochs):

    # Predictions
    y_pred = m*X + b

    # Cost Function (MSE)
    cost = (1/(2*n))*np.sum((y - y_pred)**2)

    cost_history.append(cost)

    # Gradients
    dm = (-1/n) * np.sum(X*(y - y_pred))
    db = (-1/n) * np.sum(y - y_pred)

    # Update Parameters
    m = m - learning_rate*dm
    b = b - learning_rate*db

    if i % 100 == 0:
        print(f"Epoch {i}: Cost = {cost:.4f}")
```

```python
print("Final Slope (m):",m)
print("Final Intercept (b):",b)
```

```python
y_pred = m*X + b

plt.scatter(X,y,label="Data Points")
plt.plot(X,y_pred,color='red',label="Best Fit Line")

plt.xlabel("CGPA")
plt.ylabel("Placed")
plt.title("Linear Regression using Gradient Descent")
plt.legend()

plt.show()
```

```python
plt.plot(cost_history)

plt.xlabel("Iterations")
plt.ylabel("Cost")
plt.title("Gradient Descent Convergence")

plt.show()
```

```python
mse = np.mean((y-y_pred)**2)

print("Mean Squared Error:",mse)
```

```python
ss_total = np.sum((y-np.mean(y))**2)
ss_residual = np.sum((y-y_pred)**2)

r2 = 1 - (ss_residual/ss_total)

print("R2 Score:",r2)
```

```python
print("Gradient Descent successfully minimized the cost function.")
print("A best-fit line was obtained between CGPA and placement status.")
print("Since the target variable is binary (0/1), Logistic Regression would be more suitable in practice.")
```
